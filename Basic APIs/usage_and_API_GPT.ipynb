## Basically a Sales Brochure Generator 

```python
import os
import json
from dotenv import load_dotenv
from IPython.display import Markdown, display, update_display
from my_easy_scraper import fetch_all_website_links, fetch_website_contents, fetch_website_links 
from openai import OpenAI
```

```python
load_dotenv(override=True)
api_key = os.getenv('OPENAI_API_KEY')
    
MODEL = 'gpt-4o-mini'
openai = OpenAI()
```

```python
response = openai.chat.completions.create(
    model=MODEL,
    messages=[
        {"role": "system", "content": "You are a helpful assistant."},
        {"role": "user", "content": "Write a short poem about coding."}
    ]
)
#not printing the response, since main goal was to check the usage
# print("Response:\n" + response.choices[0].message.content)
# print("\n" + "="*30 + "\n")

usage = response.usage
prompt_tokens = usage.prompt_tokens
completion_tokens = usage.completion_tokens
total_tokens = usage.total_tokens



print("--- USAGE DATA ---")
print(f"Input Tokens (Prompt): {prompt_tokens}")
print(f"Output Tokens (Completion): {completion_tokens}")
print(f"Total Tokens: {total_tokens}")
```
